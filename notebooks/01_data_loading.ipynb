```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_df = pd.read_csv("../data/raw/mitbih_train.csv", header = None)
train_df.head()
```

```python
signal = train_df.iloc[0, :-1]

plt.figure(figsize=(12,4))
plt.plot(signal)

plt.title("ECG Signal Example")
plt.xlabel("Time")
plt.ylabel("Amplitude")

plt.show()
```

```python
train_df.iloc[:, -1].unique()
```

```python
train_df.iloc[:, -1].value_counts()
```

```python
class_labels = {
    0.0: "Normal",
    1.0: "Supreventricular ectopic beat",
    2.0: "Ventricular ectopic beat",
    3.0: "Fusion beat",
    4.0: "Unknown beat"
}

plt.figure(figsize=(12,6))

for class_id, class_name in class_labels.items():
    sample = train_df[train_df.iloc[:, -1] == class_id].iloc[0, :-1]
    plt.plot(sample, label=f"Class {int(class_id)} - {class_name}")

plt.title("Example ECG Signals by Class")
plt.xlabel("Sample index")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```
